```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
import pandas as pd
import json
#from sentence_transformers import SentenceTransformer

# 1. 读取Excel文件（替换为你的Excel路径）
df = pd.read_excel("data/BSRF-HEPS-KnowledgeBase-k-v.xlsx", usecols=["k", "v"])  # 只读取k和v列

# 2. 加载SentenceTransformer模型
model = SentenceTransformer("all-MiniLM-L6-v2")

# 3. 分别对k列、v列文本向量化
k_sentences = df["k"].tolist()  # 去除空值，转为列表
v_sentences = df["v"].tolist()

k_embeddings = model.encode(k_sentences).tolist()  # 转为列表（JSON可序列化）
v_embeddings = model.encode(v_sentences).tolist()

# 4. 保存为JSON文件
with open("k_embeddings.json", "w", encoding="utf-8") as f:
    json.dump({"sentences": k_sentences, "embeddings": k_embeddings}, f, ensure_ascii=False, indent=2)

with open("v_embeddings.json", "w", encoding="utf-8") as f:
    json.dump({"sentences": v_sentences, "embeddings": v_embeddings}, f, ensure_ascii=False, indent=2)

print("向量化完成，已保存为k_embeddings.json和v_embeddings.json")
```

```python
# 4. 保存为JSON文件
with open("k_embeddings2.json", "w", encoding="utf-8") as f:
    json.dump(k_embeddings, f, ensure_ascii=False, indent=2)

with open("v_embeddings2.json", "w", encoding="utf-8") as f:
    json.dump( v_embeddings, f, ensure_ascii=False, indent=2)

print("向量化完成，已保存为k_embeddings.json和v_embeddings.json")
```
